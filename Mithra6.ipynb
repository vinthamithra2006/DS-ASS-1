```python
ds = pd.get_dummies(ds,columns=['Restaurant Name'])
ds.head()
```

```python
numeric_ds =ds.select_dtypes(include="number")
```

```python
import pandas as pd


r1 = pd.DataFrame(
    {
        "Mean": numeric_ds.mean(),
        "Median": numeric_ds.median(),
        "Mode": numeric_ds.mode().iloc[0],
    }
)


print(r1.round(3))
```

```python
r2=pd.DataFrame({
    "Variance":numeric_ds.var(),
    "Standard Deviation":numeric_ds.std()
})
print(r2.round(5))
```

```python
r3 =pd.DataFrame({
    "Minimum": numeric_ds.min(),
    "Maximum": numeric_ds.max(),
    "Range": numeric_ds.max()-numeric_ds.min()
})
print(r3.astype(int))
```

```python
r4 =pd.DataFrame({
    "Q1 (25%)":numeric_ds.quantile(0.25),
    "Q2 (Median)":numeric_ds.quantile(0.50),
    "Q3 (75%)":numeric_ds.quantile(0.75)
})
r4["IQR"]=r4["Q3 (75%)"] - r4["Q1 (25%)"]
print(r4)
```

```python
r5 =pd.DataFrame({
    "Skewness": numeric_ds.skew()
})
print(r5.round(4))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
column = numeric_ds.columns[0]
sns.histplot(numeric_ds[column],kde=True)
plt.title(f"Skewness Distribution Of {column}")
plt.xlabel("Restaurant Name_iKitchen")
plt.ylabel("Votes")
plt.show()
print("Skewness:",numeric_ds[column].skew())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
column = numeric_ds.columns[2]
sns.histplot(numeric_ds[column],kde=True)
plt.title(f"Kurtosis Distribution Of {column}")
plt.xlabel("Address")
plt.ylabel("City")
plt.show()
print("Kutrosis:",numeric_ds[column].kurt())
```

```python
Q1=numeric_ds.quantile(0.25)
Q3=numeric_ds.quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
outliers = (
    (numeric_ds > lower_limit) | (numeric_ds < upper_limit)
)
print("Lower Bound:\n", lower_limit)
print("Upper Bound:\n", upper_limit)
print("Outliers in each column:\n", numeric_ds[outliers].count())
```

```python
import pandas as pd
mean_values = numeric_ds.mean()
std_values = numeric_ds.std()
r5 = pd.DataFrame(
    {
        "Mean": mean_values,
        "Standard Deviation": std_values,
        "cv (%)": (std_values / mean_values) * 100,
    }
)
print(r5.round(4))
```

```python
ds = numeric_ds["Votes"].dropna()
mean_val = ds.mean()
median_val = ds.median()

print(f"Mean: {mean_val:.2f} | Median: {median_val:.2f}")
if mean_val > median_val:
  print("Right-skewed (Positive Skewness)")
else:
  print("Left-skewed (Negative Skewness)")
```
