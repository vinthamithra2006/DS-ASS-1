```python
hotel['children']=hotel['children'].fillna(hotel['children'].mean())
hotel['country']=hotel['country'].fillna(hotel['country'].mode()[0])
hotel['agent']=hotel['agent'].fillna(hotel['agent'].mean())
hotel['company']=hotel['company'].fillna(hotel['company'].mean())
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.figure(figsize=(12, 8))
plt.subplot(2,2,2)
plt.hist(hotel["arrival_date_year"].dropna(), bins=20, edgecolor="black")
plt.title("Histogram of Hotel")
plt.xlabel("Numerical")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns 
plt.subplot(2,2,4)
sns.boxplot(y=hotel["is_canceled"])
plt.title("Box plot of is_canceled")
plt.ylabel("Area")

plt.tight_layout()
plt.show()
```

```python
numeric_hotel = hotel.select_dtypes(include=['number'])
Q1 = numeric_hotel.quantile(0.25)
Q3 = numeric_hotel.quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
outliers = (numeric_hotel < lower_limit) | (numeric_hotel > upper_limit)
print("Lower Bound:\n", lower_limit)
print("Upper Bound:\n", upper_limit)
print("Outliers count in each column:\n", outliers.sum())
```

```python
skew = hotel['agent'].skew()
if skew > 0:
    print("Positively Skewed")
elif skew < 0:
    print("Negatively Skewed")
else:
    print("Normally Distributed")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
column = numeric_hotel.columns[0]
sns.histplot(numeric_hotel[column],kde=True)
plt.title(f"Skewness Distribution Of {column}")
plt.xlabel("Restaurant Hotel Booking")
plt.ylabel("Votes")
plt.show()
print("Skewness:",numeric_hotel[column].skew())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
column = numeric_hotel.columns[2]
sns.histplot(numeric_hotel[column],kde=True)
plt.title(f"Kurtosis Distribution Of {column}")
plt.xlabel("Restaurant Name_iKitchen")
plt.ylabel("Votes")
plt.show()
print("Skewness:",numeric_hotel[column].kurt())
```
